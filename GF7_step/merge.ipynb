```python
import os
import pandas as pd

def merge_rule8_csv(in_folder, out_path=None, pattern_suffix="Rule8_full.csv"):
    """
    将文件夹中所有以 Rule8_full.csv 结尾的文件合并为一个新的 CSV。

    参数:
    ----------
    in_folder : str
        包含待合并文件的文件夹路径
    out_path : str, optional
        输出合并文件的完整路径（默认保存在同一目录下）
    pattern_suffix : str, optional
        文件名后缀匹配模式（默认 'Rule8_full.csv'）
    """

    # 搜索匹配文件
    csv_files = [os.path.join(in_folder, f) for f in os.listdir(in_folder)
                 if f.endswith(pattern_suffix)]

    if not csv_files:
        print(f"❌ 没有找到以 '{pattern_suffix}' 结尾的文件。")
        return

    print(f"🧩 找到 {len(csv_files)} 个文件，正在合并中...")

    # 读取并合并所有 CSV
    dfs = []
    for f in csv_files:
        try:
            df = pd.read_csv(f)
            df["source_file"] = os.path.basename(f)  # ✅ 保留来源文件名
            dfs.append(df)
            print(f"  ✅ 已读取 {os.path.basename(f)} ({len(df)} 行)")
        except Exception as e:
            print(f"  ⚠️ 跳过 {os.path.basename(f)}：{e}")

    if not dfs:
        print("⚠️ 没有成功读取的文件。")
        return

    merged_df = pd.concat(dfs, ignore_index=True)

    # 输出路径
    if out_path is None:
        out_path = os.path.join(in_folder, "merged_Rule8_full.csv")

    # 保存
    merged_df.to_csv(out_path, index=False)
    print(f"\n🎉 合并完成！输出文件: {out_path}")
    print(f"总行数: {len(merged_df)}")

```
